Two Sum

```python
nums=[2,7,11,15]
target = 9
```

```python
def twoSum(nums, target):
    ls = len(nums)
    for i in range(ls):
        for j in range(i + 1, ls):
            if nums[i] + nums[j] == target:
                return[i, j]
```

```python
twoSum(nums,target)
```
