```python
nums=[2,7,11,15]
target = 18
```

```python
if nums[1] + nums[2] == target:
    print("true")
else:
    print("false")
```

```python
def two_nums():
    n = len(nums)
    for i in range(n-1):
        for j in range(i + 1 , n):
            if nums[i] + nums[j] == target:
                return [i,j]
    return []
        
```
